## 2. Data Cleaning

```python
customer.isnull().sum()
```

```python
customer['TotalCharges'].head(10)
```

```python
# Check whether any values are blank or contain only whitespace 
customer["TotalCharges"].str.strip().eq("").sum()
```

```python
# pd.to_numeric() function attempts to convert string values into numeric values
# The argument errors='coerce' converts values that cannot be interpreted as numbers to NaN
customer['TotalCharges'] = pd.to_numeric(customer['TotalCharges'], errors='coerce')
```

```python
customer['TotalCharges'].dtype
```

```python
customer["TotalCharges"].isnull().sum()
```

Investigate missing TotalCharges values

```python
customer.loc[
    customer['TotalCharges'].isnull(),
    ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']
    ]
```

```python
customer['TotalCharges'] = customer['TotalCharges'].fillna(0)
```

Data Cleaning Finding: The TotalCharges column contained 11 blank values. All 11 customers had a tenure of zero months. I converted TotalCharges into a numeric type and replaced the missing values with zero, assuming these customers had not yet accumulated charges. This assumption is documented because tenure alone does not confirm billing history.

## 4. Data Visualization

### 4.1 Overall Customer Churn

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Save the churn rates to a variable 
churn_rates = (
    customer['Churn']
    .value_counts(normalize=True)
    .reindex(['Yes', 'No'])
    * 100
)

# Create bar chart of Customer Churn 
plt.figure(figsize=(7, 5))

bars = plt.bar(
    churn_rates.index,
    churn_rates.values
)

plt.bar_label(
    bars,
    fmt="%.1f%%",
    padding=3
)

plt.title("Overall Customer Churn Rate")
plt.xlabel("Customer Churn Status")
plt.ylabel("Percentage of Customers (%)")
plt.ylim(0, 100)

# Save the visualize into the visualizations folder 
plt.savefig(
    "visualizations/overall_churn.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.clf()
```

### 4.2 Churn by Contract Type 

Step 1: Prepare the data 

```python
# Create a variable that gives the percentage of churn per contract
contract_churn = (
    customer.groupby('Contract')['Churn']
    .value_counts(normalize=True)
    .unstack()
    ['Yes']
    * 100
)
```

Step 2: Order the contract categories 

```python
contract_churn = contract_churn.reindex([
    'Month-to-month',
    'One year',
    'Two year'
])
```

Step 3: Create the bar chart

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(
    contract_churn.index,
    contract_churn.values
)

plt.title('Customer Churn Rate by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate (%)')
plt.ylim(0, 50)

plt.bar_label(
    bars, 
    fmt='%.1f%%',
    padding=3
)

plt.savefig(
    "visualizations/churn_by_contract.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.clf()
```

### 4.3 Customer Churn by Tenure 

Step 1: Prepare the data

```python
customer['TenureGroup'].value_counts().sort_index()
```

```python
# Calculate the percentage of customers who churned within each tenure group
tenure_churn = (
    customer.groupby('TenureGroup')['Churn']
    .value_counts(normalize=True)
    .unstack(fill_value=0)
    ['Yes']
    * 100
)
tenure_churn
```

Step 2: Visualize the churn rates

```python
plt.figure(figsize=(9, 5))

bars = plt.bar(
    tenure_churn.index,
    tenure_churn.values
)

plt.title('Customer Churn rate by Tenure')
plt.xlabel('Customer Tenure')
plt.ylabel('Churn Rate (%)')

plt.ylim(0, 60)

plt.bar_label(
    bars,
    fmt='%.1f%%',
    padding=3
)

plt.savefig(
    "visualizations/churn_by_tenure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
