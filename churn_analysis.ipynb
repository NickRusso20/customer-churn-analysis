## 2. Data Cleaning

```python
customer.isnull().sum()
```

```python
customer['TotalCharges'].head(10)
```

```python
# Check whether any values are blank or contain only whitespace 
customer["TotalCharges"].str.strip().eq("").sum()
```

```python
# pd.to_numeric() function attempts to convert string values into numeric values
# The argument errors='coerce' converts values that cannot be interpreted as numbers to NaN
customer['TotalCharges'] = pd.to_numeric(customer['TotalCharges'], errors='coerce')
```

```python
customer['TotalCharges'].dtype
```

```python
customer["TotalCharges"].isnull().sum()
```

Investigate missing TotalCharges values

```python
customer.loc[
    customer['TotalCharges'].isnull(),
    ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']
    ]
```

```python
customer['TotalCharges'] = customer['TotalCharges'].fillna(0)
```

Data Cleaning Finding: The TotalCharges column contained 11 blank values. All 11 customers had a tenure of zero months. I converted TotalCharges into a numeric type and replaced the missing values with zero, assuming these customers had not yet accumulated charges. This assumption is documented because tenure alone does not confirm billing history.
